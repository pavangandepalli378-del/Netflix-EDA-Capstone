````python
import pandas as pd

df = pd.read_csv('/content/NETFLIX MOVIES AND TV SHOWS CLUSTERING.csv')

df.head()
````

````python
# Check number of rows and columns
df.shape
````

````python
# Check column names
df.columns
````

````python
# Check data types and non-null values
df.info()
````

````python
# Count missing values in each column
df.isnull().sum()
````

````python
# Missing-value percentage
(df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
````

````python
df.describe(include='all').T
````

````python
# Handle missing values in text columns

df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
````

````python
# Fill missing date_added and rating with their respective modes

df['date_added'] = df['date_added'].fillna(df['date_added'].mode()[0])
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
````

````python
df.isnull().sum()
````

````python
print("Number of duplicate rows:", df.duplicated().sum())
````

````python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
````

````python
df.isnull().sum()
````

````python
df.duplicated().sum()
````

````python
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# Chart 1: Distribution of Movies and TV Shows

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='type')

plt.title('Distribution of Movies and TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.show()
````

````python
# Chart 2: Distribution of Netflix content ratings

plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x='rating',
    order=df['rating'].value_counts().index
)

plt.title('Distribution of Netflix Content Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
````

````python
# Chart 3: Distribution of content by release year

plt.figure(figsize=(12, 6))

sns.histplot(
    data=df,
    x='release_year',
    bins=30,
    kde=True
)

plt.title('Distribution of Netflix Content by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
````

````python
# Chart 4: Top 10 countries by number of Netflix titles

country_counts = (
    df['country']
    .fillna('Unknown')
    .str.split(',')
    .explode()
    .str.strip()
)

country_counts = country_counts[country_counts != 'Unknown'].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=country_counts.values,
    y=country_counts.index
)

plt.title('Top 10 Countries by Number of Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
````

````python
# Chart 5: Top 10 genres on Netflix

genre_counts = (
    df['listed_in']
    .fillna('Unknown')
    .str.split(',')
    .explode()
    .str.strip()
)

genre_counts = genre_counts[genre_counts != 'Unknown'].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=genre_counts.values,
    y=genre_counts.index
)

plt.title('Top 10 Genres on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
````

````python
# Chart 6: Movie duration distribution

movie_df = df[df['type'] == 'Movie'].copy()

movie_df['duration_minutes'] = (
    movie_df['duration']
    .str.extract(r'(\d+)')[0]
    .astype(float)
)

plt.figure(figsize=(10, 6))

sns.histplot(
    data=movie_df,
    x='duration_minutes',
    bins=30,
    kde=True
)

plt.title('Distribution of Movie Durations')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Number of Movies')
plt.show()
````

````python
# Chart 7: Distribution of TV show seasons

tv_df = df[df['type'] == 'TV Show'].copy()

tv_df['seasons'] = (
    tv_df['duration']
    .str.extract(r'(\d+)')[0]
    .astype(float)
)

plt.figure(figsize=(10, 6))

sns.countplot(
    data=tv_df,
    x='seasons',
    order=sorted(tv_df['seasons'].dropna().unique())
)

plt.title('Distribution of TV Shows by Number of Seasons')
plt.xlabel('Number of Seasons')
plt.ylabel('Number of TV Shows')
plt.show()
````

````python
# Chart 8: Netflix content added by year

df['date_added_clean'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)

added_year = df['date_added_clean'].dt.year.value_counts().sort_index()

plt.figure(figsize=(12, 6))

sns.lineplot(
    x=added_year.index,
    y=added_year.values,
    marker='o'
)

plt.title('Netflix Content Added by Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles Added')
plt.show()
````

````python
# Chart 9: Content added by year and content type

year_type = (
    df.dropna(subset=['date_added_clean'])
      .groupby([df['date_added_clean'].dt.year, 'type'])
      .size()
      .reset_index(name='count')
)

year_type.columns = ['year', 'type', 'count']

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=year_type,
    x='year',
    y='count',
    hue='type',
    marker='o'
)

plt.title('Netflix Content Added by Year and Type')
plt.xlabel('Year')
plt.ylabel('Number of Titles Added')
plt.show()
````

````python
# Chart 10: Ratings by content type

top_ratings = df['rating'].value_counts().head(8).index

rating_type = df[df['rating'].isin(top_ratings)]

plt.figure(figsize=(12, 6))

sns.countplot(
    data=rating_type,
    x='rating',
    hue='type',
    order=top_ratings
)

plt.title('Content Ratings by Content Type')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
````

````python
# Chart 11: Top countries by content type

country_type = df[['country', 'type']].dropna().copy()

country_type['country'] = country_type['country'].str.split(',')
country_type = country_type.explode('country')

country_type['country'] = country_type['country'].str.strip()

top_countries = country_type['country'].value_counts().head(10).index

country_type = country_type[
    country_type['country'].isin(top_countries)
]

plt.figure(figsize=(12, 6))

sns.countplot(
    data=country_type,
    y='country',
    hue='type',
    order=top_countries
)

plt.title('Top 10 Countries by Netflix Content Type')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
````

````python
# Chart 12: Top genres by content type

genre_type = df[['listed_in', 'type']].dropna().copy()

genre_type['listed_in'] = genre_type['listed_in'].str.split(',')
genre_type = genre_type.explode('listed_in')

genre_type['listed_in'] = genre_type['listed_in'].str.strip()

top_genres = genre_type['listed_in'].value_counts().head(10).index

genre_type = genre_type[
    genre_type['listed_in'].isin(top_genres)
]

plt.figure(figsize=(12, 7))

sns.countplot(
    data=genre_type,
    y='listed_in',
    hue='type',
    order=top_genres
)

plt.title('Top 10 Genres by Content Type')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
````

````python
# Chart 13: Movie duration by rating

movie_rating = df[df['type'] == 'Movie'].copy()

movie_rating['duration_minutes'] = (
    movie_rating['duration']
    .str.extract(r'(\d+)')[0]
    .astype(float)
)

top_movie_ratings = movie_rating['rating'].value_counts().head(8).index

movie_rating = movie_rating[
    movie_rating['rating'].isin(top_movie_ratings)
]

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=movie_rating,
    x='rating',
    y='duration_minutes'
)

plt.title('Movie Duration by Content Rating')
plt.xlabel('Rating')
plt.ylabel('Duration (Minutes)')
plt.xticks(rotation=45)
plt.show()
````

````python
# Chart 14: Release decade by content type

df['release_decade'] = (df['release_year'] // 10) * 10

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x='release_decade',
    hue='type'
)

plt.title('Netflix Content by Release Decade and Type')
plt.xlabel('Release Decade')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
````

````python
# Chart 15: Top 10 Directors

director_data = df['director'].dropna()

director_data = director_data[director_data != 'Unknown']

director_counts = (
    director_data
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=director_counts.values,
    y=director_counts.index
)

plt.title('Top 10 Directors by Number of Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Director')
plt.show()
````

````python
# Chart 16: Title Length by Content Type

df['title_length'] = df['title'].fillna('').str.len()

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='type',
    y='title_length'
)

plt.title('Title Length by Content Type')
plt.xlabel('Content Type')
plt.ylabel('Title Length (Characters)')
plt.show()
````

````python
# Chart 17: Content Added by Month

df['date_added_clean'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)

df['added_month'] = df['date_added_clean'].dt.month_name()

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

plt.figure(figsize=(12, 5))

sns.countplot(
    data=df.dropna(subset=['added_month']),
    x='added_month',
    order=month_order
)

plt.title('Netflix Content Added by Month')
plt.xlabel('Month')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
````

````python
# Chart 18: Ratings by Release Decade

df['release_decade'] = (df['release_year'] // 10) * 10

top_ratings = df['rating'].value_counts().head(8).index

rating_decade = pd.crosstab(
    df.loc[df['rating'].isin(top_ratings), 'rating'],
    df.loc[df['rating'].isin(top_ratings), 'release_decade']
)

plt.figure(figsize=(12, 6))

sns.heatmap(
    rating_decade,
    annot=True,
    fmt='d'
)

plt.title('Netflix Ratings by Release Decade')
plt.xlabel('Release Decade')
plt.ylabel('Rating')
plt.show()
````

````python
# Chart 19: Top Countries by Rating

country_rating = df[['country', 'rating']].dropna().copy()

country_rating['country'] = country_rating['country'].str.split(',')
country_rating = country_rating.explode('country')
country_rating['country'] = country_rating['country'].str.strip()

top_countries = country_rating['country'].value_counts().head(8).index
top_ratings = country_rating['rating'].value_counts().head(6).index

filtered_data = country_rating[
    country_rating['country'].isin(top_countries) &
    country_rating['rating'].isin(top_ratings)
]

country_rating_table = pd.crosstab(
    filtered_data['country'],
    filtered_data['rating']
)

plt.figure(figsize=(12, 6))

sns.heatmap(
    country_rating_table,
    annot=True,
    fmt='d'
)

plt.title('Top Countries by Netflix Rating')
plt.xlabel('Rating')
plt.ylabel('Country')
plt.show()
````

````python
# Chart 20: Top Genres by Release Decade

genre_data = df[['listed_in', 'release_decade']].dropna().copy()

genre_data['genre'] = genre_data['listed_in'].str.split(',')
genre_data = genre_data.explode('genre')
genre_data['genre'] = genre_data['genre'].str.strip()

top_genres = genre_data['genre'].value_counts().head(8).index

filtered_genres = genre_data[
    genre_data['genre'].isin(top_genres)
]

genre_decade_table = pd.crosstab(
    filtered_genres['genre'],
    filtered_genres['release_decade']
)

recent_decades = sorted(genre_decade_table.columns)[-10:]

genre_decade_table = genre_decade_table[recent_decades]

plt.figure(figsize=(14, 7))

sns.heatmap(
    genre_decade_table,
    annot=True,
    fmt='d'
)

plt.title('Top Netflix Genres by Release Decade')
plt.xlabel('Release Decade')
plt.ylabel('Genre')
plt.show()
````
